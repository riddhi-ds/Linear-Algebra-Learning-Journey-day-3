📏**Norms**

1. *A norm is a mathematical function used to measure the length or magnitude of a vector.*
2. *Just as we measure the distance between two points, norms help us measure how large a vector is.*

```python
import numpy as np

v = np.array([3,4])

print("Vector:", v)
```

**Why are Norms Important?**

Norms are widely used in:

- *Machine Learning*
- *Data Science*
- *Optimization*
- *Deep Learning*
- *Recommendation Systems*
- *Computer Vision*

*They help measure: Distance , Similarity , Error , Model Complexity*

**📏L1 Norm (Manhattan Distance)**

*The L1 norm is the sum of the absolute values of all vector elements.*

```python
l1_norm = np.linalg.norm(v , ord=1)

print("L1 Norm:", l1_norm)
```

**📏L2 Norm (Euclidean Distance)**

*The L2 norm measures the straight-line distance from the origin to the vector.*

```python
l2_norm = np.linalg.norm(v , ord=2)

print("L2 Norm:", l2_norm)
```

**📏Infinity Norm (Maximum Norm)**

*The infinity norm returns the largest absolute value in the vector.*

```python
inf_norm = np.linalg.norm(v , ord=np.inf)

print("Infinity Norm:", inf_norm)
```

**📏 Unit Vectors**

1. *A unit vector is a vector whose magnitude (length) is exactly 1.*

2. *It represents direction only, without changing the vector's orientation.*

Eg:

[3, 4]  -> Original Vector

[0.6, 0.8] -> Unit Vector

*Both point in the same direction, but the second vector has magnitude 1.*

**Why Do We Need Unit Vectors?**

Unit vectors are used in:

- *Machine Learning*
- *Deep Learning*
- *Computer Graphics*
- *Physics*
- *Cosine Similarity*

**They allow us to compare directions independently of magnitude.**

```python
# Create vector
vector = np.array([2, -5, 8])

# Find Magnitude
magnitude = np.linalg.norm(vector)

# Convert to Unit Vector
unit_vector = vector / magnitude

print("Original Vector:", vector)
print("\nMagnitude:", magnitude)
print("\nUnit Vector:", unit_vector)
```

```python
#Verify Magnitude is 1

print(np.linalg.norm(unit_vector))  # This confirms it is a unit vector.
```

**📏 Generalized Lp Norm**

1. *The Lp Norm is a general formula that defines a whole family of norms.*

2. *The L1 Norm, L2 Norm, and Infinity Norm are all special cases of the Lp Norm.*

3. *Instead of having separate formulas, we can use one generalized formula.*

```python
import numpy as np

v = np.array([1, 2, 3])

print("L1 :", np.linalg.norm(v, ord=1))
print("L2 :", np.linalg.norm(v, ord=2))
print("L3 :", np.linalg.norm(v, ord=3))
print("L_infinity :", np.linalg.norm(v, ord=np.inf))
```

**💡 Key Takeaways**

- *Lp Norm is the generalized form of all norms.*

- *L1, L2, and Infinity Norm are special cases.*

- *The value of p determines how the vector magnitude is measured.*

- *Lp norms are heavily used in Machine Learning, Optimization, and Regularization.*

**📏 Basis Vectors**

- *A basis is a set of vectors that can be used to represent every vector in a vector space.*

- *The vectors in this set are called basis vectors.*

- *Think of basis vectors as the building blocks of all other vectors.*

**Why Are Basis Vectors Important?**

Basis vectors help us:

- *Describe coordinates*

-  *Build vector spaces*

-  *Understand transformations*

- *Learn PCA and dimensionality reduction*

- *Understand embeddings in Machine Learning*

```python
# Create Basis Vector


# 2 D
import numpy as np

e1 = np.array([1, 0])
e2 = np.array([0, 1])

print("e1:", e1)
print("e2:", e2)

v = 3 * e1 + 5 * e2

print(v)
```

```python
# 3 D

e1 = np.array([1,0,0])
e2 = np.array([0,1,0])
e3 = np.array([0,0,1])

print(e1)
print(e2)
print(e3)
```

**📏Orthogonal Vectors**

*Two vectors are called orthogonal if they are perpendicular to each other (90° angle).*

*In Linear Algebra, instead of measuring the angle directly, we use the dot product.*

a · b = 0
-> where · represents the dot product.

For two vectors:
-> a = [a₁, a₂]
  b = [b₁, b₂]
The dot product is:  
-> a · b = a₁b₁ + a₂b₂

**Why Are Orthogonal Vectors Important?**

Orthogonal vectors are used in:

- *PCA (Principal Component Analysis)*

Features are transformed into orthogonal directions.

- *Machine Learning*

Helps remove redundancy between features.

- *Singular Value Decomposition (SVD)*

Built using orthogonal vectors.

- *Deep Learning*

Used in matrix decompositions and optimization.

```python
import numpy as np

a = np.array([1, 1])
b = np.array([1, -1])

dot_product = np.dot(a, b)

print(dot_product)
```

**💡 Key Takeaways**

-  *Orthogonal vectors are perpendicular.*

-  *Their dot product equals 0.*

-  *Dot Product is the easiest way to check orthogonality.*

-  *Orthogonality is a core concept behind PCA, SVD, dimensionality reduction, and many Machine Learning algorithms.*

**📏Matrices**

- *A matrix is a rectangular arrangement of numbers organized into rows and columns.*

- *A matrix is essentially a collection of numbers stored in a tabular format.*

**Why Are Matrices Important?**

Matrices are used everywhere in:

- *Data Science*

- *Machine Learning*

- *Deep Learning*

- *Computer Vision*

- *Recommendation Systems*

*Almost all ML algorithms work with data represented as matrices.*

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

print(A)
```

```python
print(A.shape)
```

```python
B = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(B.shape)
```

```python
# Matrix Addition

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print(A + B)
```

```python
# Matrix Subtraction

print(B - A)
```

```python
# Matrix Transpose

print(A.T)
```

```python
# Matrix Multiplication

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print(A @ B)
```

**💡 Key Takeaways**

- *Matrix = rows × columns*

-  *Shape tells dimensions*

-  *Addition/Subtraction require same shape*

-  *Transpose swaps rows and columns*

-  *Matrix multiplication is fundamental in ML and Deep Learning*

-  *Most datasets are stored as matrices*

**📏Generic Tensor Notation**

*A tensor is a generalized mathematical object that extends scalars, vectors, and matrices.*

*Think of tensors as containers for numerical data with different numbers of dimensions.*

Scalar  -> Rank 0 Tensor

Vector  -> Rank 1 Tensor

Matrix  -> Rank 2 Tensor

Tensor  -> Rank 3 or Higher

```python
import numpy as np

# Rank 0 Tensor - Scalar
scalar = np.array(10)

# Rank 1 Tensor -  Vector
vector = np.array([1, 2, 3])

# Rank 2 Tensor - Matrix
matrix = np.array([[1, 2], [3, 4]])

print(scalar.ndim)
print(vector.ndim)
print(matrix.ndim)
```

**📏Higher-Rank Tensors**

*Any tensor with rank 3 or more is called a higher-rank tensor.*

These are heavily used in:

- *Deep Learning*
- *Computer Vision*
- *NLP*
- *Generative AI*

**Why Are Tensors Important?**

Frameworks like:

- *NumPy*
- *TensorFlow*
- *PyTorch*

store and process data using tensors.

*Every image, video, embedding, and neural network input is represented as a tensor.*

📏**Image Tensor Example**

*A color image consists of:*
*Height × Width × Channels*

```python
image = np.random.rand(224, 224, 3)

print(image.shape)

# 224 = Height
# 224 = Width
# 3   = RGB Channels
```

**📏Video Tensor Example**

*A video is a collection of image frames.*
*Frames × Height × Width × Channels*

```python
video = np.random.rand(10, 224, 224, 3)

print(video.shape)

# 10  = Frames
# 224 = Height
# 224 = Width
# 3   = RGB Channels
```

**💡 Key Takeaways**

- *Scalar = Rank 0 Tensor*

-  *Vector = Rank 1 Tensor*

-  *Matrix = Rank 2 Tensor*

-  *Tensor = Rank 3 or Higher*

-  *Rank = Number of Dimensions (Axes)*

-  *Images and videos are represented using higher-rank tensors*

- *Tensors are the fundamental data structure in Deep Learning*
